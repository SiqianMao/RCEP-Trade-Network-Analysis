```python
base_dir = "/content/drive/MyDrive/RCEP_Trade_Network_Project"

clean_dir = base_dir + "/02_Data_Clean"
gephi_dir = base_dir + "/03_Gephi"
```

```python
years = [2022, 2023, 2024]

for year in years:
    print("Processing:", year)

    edge_path = f"{clean_dir}/RCEP_EdgeList_{year}_Named.csv"
    metrics_path = f"{clean_dir}/RCEP_Network_Metrics_{year}.csv"

    edges = pd.read_csv(edge_path)
    metrics = pd.read_csv(metrics_path)

    G = nx.from_pandas_edgelist(
        edges,
        source="Source",
        target="Target",
        edge_attr="Weight",
        create_using=nx.DiGraph()
    )

    # 给节点加入指标属性
    for _, row in metrics.iterrows():
        country = row["Country"]
        if country in G.nodes:
            G.nodes[country]["WeightedOutDegree"] = row["WeightedOutDegree"]
            G.nodes[country]["WeightedInDegree"] = row["WeightedInDegree"]
            G.nodes[country]["Betweenness"] = row["Betweenness"]
            G.nodes[country]["Eigenvector"] = row["Eigenvector"]
            G.nodes[country]["Degree"] = row["Degree"]
            G.nodes[country]["Year"] = year

    output_path = f"{gephi_dir}/RCEP_Gephi_{year}.gexf"
    nx.write_gexf(G, output_path)

    print("Saved to:", output_path)
```
